```python
def square(x):
    return x**2
```

```python
def custom_function(x):
    if x>=2:
        return 1
    else:
        return 0
```

```python
def chech_continuity(f, a):
    """
    Checks if the given functions is continuous at any point "a".
    """
    
    left_hand_limit = f(a-0.00001)
    right_hand_limit = f(a+0.00001)
    function_at_a = f(a)
    
    print(left_hand_limit, right_hand_limit, function_at_a)
    
```

```python
chech_continuity(square,1)
```

```python
chech_continuity(custom_function,2)
```
