```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
def plot_function(func):
    x_val = np.linspace(-10,10,1000)
    y_val = [func(x) for x in x_val]
    plt.plot(x_val, y_val)
```

```python
plot_function(lambda x:x**2)
```

```python
def derivative(f, x):
    delta_x = 0.000001
    return (f(x+delta_x) - f(x))/delta_x
    
```

```python
derivative(lambda x:x**2, 1)
```

```python
derivative(lambda x:x**2, -3)
```

```python
def plot_derivative(f):
    delta_x = 0.00000001
    x_val = np.linspace(-10,10,1000)
    y_val = (f(x_val + delta_x) - f(x_val))/delta_x
    plt.plot(x_val, y_val)
```

```python
plot_derivative(lambda x:x**2)
```

```python
plot_function(lambda x:x**2)
plot_derivative(lambda x:x**2)
```

```python
plot_function(lambda x:x**3)
plot_derivative(lambda x:x**3)
```

```python
plot_function(lambda x:np.exp(x))
plot_derivative(lambda x:np.exp(x))
```

```python
plot_function(lambda x:np.abs(x))
plot_derivative(lambda x:np.abs(x))
```

```python
plot_function(lambda x:np.sin(x))
plot_derivative(lambda x:np.sin(x))
```
